## 7. Regrouper avec `groupby`

## 8. Visualiser

### 🎨 Comment pandas dessine un graphique

Les graphiques sont dessinés par **matplotlib**, la bibliothèque de graphiques de référence en Python. On l'a importée au début sous le nom `plt` : dans ce notebook, pandas l'appelle pour nous, on n'a donc pas besoin d'écrire `plt` nous-mêmes. Tu le rencontreras plus tard pour personnaliser des graphiques plus complexes. pandas fournit un raccourci : `.plot`, qui se met **au bout** d'une colonne ou d'un résultat, puis le **type** de graphique :

```python
donnees.plot.bar(title="Mon titre", xlabel="axe horizontal", ylabel="axe vertical", rot=0);
#  ↑ ce qu'on veut dessiner   ↑ le type     ↑ les réglages (tous facultatifs)
```

**Quel graphique pour quelle question ?**

| Question | Type | Code |
|---|---|---|
| Comparer des **catégories** (une valeur par espèce) | barres | `.plot.bar()` |
| Voir comment les valeurs d'**une** colonne se **répartissent** | histogramme | `.plot.hist(bins=...)` |
| Voir le **lien entre deux** colonnes | nuage de points | `.plot.scatter(x="...", y="...")` |
| Suivre une **évolution** dans le temps | courbe | `.plot.line()` (pas utilisé ici) |

**Les réglages les plus utiles**

| Réglage | Effet |
|---|---|
| `title="..."` | titre du graphique |
| `xlabel="..."`, `ylabel="..."` | nom des axes : un graphique sans unité est difficile à lire |
| `rot=0` | écrit les étiquettes de l'axe horizontal à l'horizontale (sinon elles sont penchées) |
| `figsize=(8, 4)` | taille du graphique : (largeur, hauteur) |
| `color="..."` | **une seule** couleur pour toutes les barres ou tous les points, par exemple `"tab:green"` ([liste des couleurs](https://matplotlib.org/stable/gallery/color/named_colors.html)) |
| `bins=20` | (histogramme) nombre de barres |

**Lire un histogramme** : l'axe horizontal découpe les valeurs en intervalles (les `bins`), et la **hauteur de chaque barre** est le **nombre de fleurs** dont la mesure tombe dans cet intervalle. Une barre haute signifie beaucoup de fleurs de cette taille.

Le **point-virgule `;`** en fin de ligne n'est pas obligatoire : il évite seulement qu'un texte technique (`<Axes: …>`) s'affiche au-dessus du graphique.

### 📖 Exemple — diagramme en barres
On calcule d'abord la moyenne par espèce avec `groupby` (partie 7), puis on met `.plot.bar(...)` au bout.

````python
iris.groupby("espece")["longueur_petale"].mean().plot.bar(
    title="Longueur moyenne du pétale", ylabel="cm", rot=0);
````

### 📖 Exemple — nuage de points
Chaque fleur devient un point : sa position horizontale est sa longueur de pétale (`x`), sa position verticale sa largeur (`y`).

Pour colorer les points selon l'espèce, on crée d'abord une colonne de couleurs avec `map`, qui remplace chaque valeur par une autre grâce à un dictionnaire : `"setosa"` devient `"tab:blue"`, etc. On la donne ensuite au réglage `c=` : contrairement à `color=`, qui applique **une seule** couleur à tous les points, `c=` reçoit **une couleur par point**, ici selon l'espèce de chaque fleur.

````python
couleurs = iris["espece"].map({"setosa": "tab:blue", "versicolor": "tab:orange", "virginica": "tab:green"})
iris.plot.scatter(x="longueur_petale", y="largeur_petale", c=couleurs,
                  title="Pétales : longueur vs largeur");
````
